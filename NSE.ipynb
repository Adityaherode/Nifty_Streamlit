```python
import yfinance as yf
import pandas as pd
import datetime as dt
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
s=dt.datetime(2025,1,1)
e=dt.datetime(2026,2,5)
```

```python
tcs=yf.download('tcs.ns', start=s, end=e)
tcs.columns=tcs.columns.get_level_values(0)
tcs=tcs.reset_index()
tcs['stock']='TCS'
tcs
```

```python
infy=yf.download('INFY.NS', start=s, end=e)
infy.columns=infy.columns.get_level_values(0)
infy=infy.reset_index()
infy['stock']='INFY'
infy
```

```python
hcltech=yf.download('HCLTECH.NS', start=s, end=e)
hcltech.columns=hcltech.columns.get_level_values(0)
hcltech=hcltech.reset_index()
hcltech['stock']='HCLTECH'
hcltech
```

```python
df=pd.concat([tcs,infy,hcltech],axis=0)
```

```python
wipro=yf.download('WIPRO.ns', start=s, end=e)
wipro.columns=wipro.columns.get_level_values(0)
wipro=wipro.reset_index()
wipro['stock']='WIPRO'
wipro
```

```python
ltim=yf.download('LTIM.ns', start=s, end=e)
ltim.columns=ltim.columns.get_level_values(0)
ltim=ltim.reset_index()
ltim['stock']='LTIM'
ltim
```

```python
techm=yf.download('TECHM.ns', start=s, end=e)
techm.columns=techm.columns.get_level_values(0)
techm=techm.reset_index()
techm['stock']='TECHM'
techm
```

```python
persistent=yf.download('PERSISTENT.ns', start=s, end=e)
persistent.columns=persistent.columns.get_level_values(0)
persistent=persistent.reset_index()
persistent['stock']='PERSISTENT'
persistent
```

```python
coforge=yf.download('COFORGE.ns', start=s, end=e)
coforge.columns=coforge.columns.get_level_values(0)
coforge=coforge.reset_index()
coforge['stock']='COFORGE'
coforge
```

```python
mphasis=yf.download('MPHASIS.ns', start=s, end=e)
mphasis.columns=mphasis.columns.get_level_values(0)
mphasis=mphasis.reset_index()
mphasis['stock']='MPHASIS'
mphasis
```

```python
ltts=yf.download('LTTS.ns', start=s, end=e)
ltts.columns=ltts.columns.get_level_values(0)
ltts=ltts.reset_index()
ltts['stock']='LTTS'
ltts
```

```python
df=pd.concat([tcs,infy,hcltech,techm,wipro,ltim,persistent,coforge,mphasis,ltts], axis=0)
```

```python
df
```

```python
df.to_csv("Stock_NSE.csv")
```
